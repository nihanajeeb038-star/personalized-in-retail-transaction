```python
import pandas as pd

data = {
    'item': [
        'T-Shirt','Jeans','Jacket','Sneakers',
        'Laptop','Smartphone','Headphones',
        'Watch','Backpack','Sunglasses'
    ],
    'tags': [
        'clothing casual cotton fashion',
        'clothing denim casual fashion',
        'clothing winter fashion warm',
        'footwear casual sports fashion',
        'electronics computer work tech',
        'electronics mobile communication tech',
        'electronics audio music tech',
        'accessory fashion wearable time',
        'accessory bag travel fashion',
        'accessory fashion summer style'
    ],
    'price': [500,1200,2500,2000,60000,30000,2500,1500,1800,800],
    'rating': [4.2,4.5,4.3,4.4,4.8,4.7,4.3,4.2,4.1,4.0]
}

df = pd.DataFrame(data)
df
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

cv = CountVectorizer()
vectors = cv.fit_transform(df['tags']).toarray()
similarity = cosine_similarity(vectors)

def recommend(item):
    index = df[df['item'] == item].index[0]
    distances = similarity[index]

    items_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]

    return [(df.iloc[i[0]], i[1]) for i in items_list]
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML

# Title
display(HTML("<h1 style='color:purple;'>🛒 Retail Shopping Dashboard</h1>"))

dropdown = widgets.Dropdown(
    options=df['item'].tolist(),
    description='Product:',
    layout=widgets.Layout(width='50%')
)

button = widgets.Button(
    description="Show Similar Products",
    button_style='warning'
)

output = widgets.Output()

def show_cards(results):
    html = "<div style='display:flex; flex-wrap:wrap;'>"

    for row, score in results:
        html += f"""
        <div style="
            border:1px solid #ddd;
            border-radius:10px;
            padding:15px;
            margin:10px;
            width:200px;
            box-shadow:2px 2px 10px #ccc;
            text-align:center;
        ">
            <h4>{row['item']}</h4>
            <p>💲 {row['price']}</p>
            <p>⭐ {row['rating']}</p>
            <p style='color:green;'>Match: {round(score,2)}</p>
        </div>
        """

    html += "</div>"
    display(HTML(html))

def on_click(b):
    with output:
        clear_output()

        item = dropdown.value
        results = recommend(item)

        display(HTML(f"<h3>🛍️ Similar Products for {item}</h3>"))
        show_cards(results)

button.on_click(on_click)

display(dropdown, button, output)
```
